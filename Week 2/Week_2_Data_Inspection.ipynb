```python
import pandas as pd
import numpy as np

url = "https://github.com/Mithuinps/Moneyball-IPL-Player-Valuation./raw/refs/heads/main/IPL_2008_2026_Merged.xlsx"

df = pd.read_excel(url)

print("Dataset loaded successfully!")
```

```python
df.head()
```

```python
print("Total Rows:", df.shape[0])
print("Total Columns:", df.shape[1])
```

```python
print("IPL Dataset Column Names:\n")

for column in df.columns:
    print(column)
```

```python
df.info()
```

```python
print("Missing Values in Each Column:\n")
print(df.isnull().sum())
```

```python
print("Total Duplicate Rows:", df.duplicated().sum())
```

```python
report = pd.DataFrame({
    "Column": df.columns,
    "Data_Type": df.dtypes.astype(str).values,
    "Missing_Values": df.isnull().sum().values,
    "Missing_Percentage": (df.isnull().mean().values * 100).round(2),
    "Unique_Values": df.nunique().values
})

print("IPL DATASET - FIRST IMPRESSION REPORT")
print("Total Rows:", df.shape[0])
print("Total Columns:", df.shape[1])
print("Duplicate Rows:", df.duplicated().sum())

display(report)
```

```python
report.to_csv("Week2_First_Impression_Report.csv", index=False)

print("First Impression Report saved successfully!")
```

```python
summary = pd.DataFrame({
    "Metric": [
        "Total Rows",
        "Total Columns",
        "Duplicate Rows",
        "Total Missing Values",
        "Columns with Missing Values"
    ],
    "Value": [
        df.shape[0],
        df.shape[1],
        df.duplicated().sum(),
        df.isnull().sum().sum(),
        (df.isnull().sum() > 0).sum()
    ]
})

display(summary)

summary.to_csv("Week2_Dataset_Summary.csv", index=False)
```
