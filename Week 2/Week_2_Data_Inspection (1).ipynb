```python
report = pd.DataFrame({
    "Column": df.columns,
    "Data_Type": df.dtypes.astype(str).values,
    "Missing_Values": df.isnull().sum().values,
    "Missing_Percentage": (df.isnull().mean().values * 100).round(2),
    "Unique_Values": df.nunique().values
})

print("IPL DATASET - FIRST IMPRESSION REPORT")
print("Total Rows:", df.shape[0])
print("Total Columns:", df.shape[1])
print("Duplicate Rows:", df.duplicated().sum())

display(report)
```

```python
import pandas as pd

df_auction = pd.read_parquet("/content/auction_all_years.parquet")

print("Dataset loaded successfully!")
print("Dataset shape:", df_auction.shape)

print("\nColumn names:")
print(df_auction.columns.tolist())

display(df_auction.head())
```

```python
print("First Auction Year:", df_auction["year"].min())
print("Last Auction Year:", df_auction["year"].max())

print("\nRecords per year:")
display(df_auction["year"].value_counts().sort_index())
```

```python
print("Available Auction Years:")
print(sorted(df_auction["year"].unique()))

print("\nMissing Values:")
print(df_auction.isnull().sum())

print("\nAuction Status:")
print(df_auction["status"].value_counts())

print("\nSample Player Names:")
display(df_auction[["year", "player_name", "team", "final_price_lakh", "status"]].head(10))
```

```python
print("Auction records for each year:")

print(
    df_auction.groupby("year").size().to_string()
)

print("\nRecords in 2024:")

display(
    df_auction[df_auction["year"] == 2024]
)
```

```python
print("2024 records:", len(df_auction[df_auction["year"] == 2024]))

print("\nAvailable years:")
print(sorted(df_auction["year"].unique()))

print("\nRecords by acquisition type:")
print(df_auction["acquisition_type"].value_counts())

print("\nRecords by source:")
print(df_auction["source"].value_counts())
```

```python
print("Auction Price Summary")

print("\nBase Price Missing:")
print(df_auction["base_price_lakh"].isnull().sum())

print("\nFinal Price Missing:")
print(df_auction["final_price_lakh"].isnull().sum())

print("\nAuction Status:")
print(df_auction["status"].value_counts())

print("\nSample Auction Prices:")
display(
    df_auction[
        ["year", "player_name", "team",
         "base_price_lakh", "final_price_lakh", "status"]
    ].head(15)
)
```

```python
print("Total Auction Records:", len(df_auction))

print("\nBase Price Missing:")
print(df_auction["base_price_lakh"].isnull().sum())

print("\nFinal Price Missing:")
print(df_auction["final_price_lakh"].isnull().sum())

print("\nFinal Price Available by Status:")
print(
    df_auction.groupby("status")["final_price_lakh"]
    .agg(["count", "mean"])
)

print("\nRecords with Missing Final Price:")
display(
    df_auction[
        df_auction["final_price_lakh"].isnull()
    ][
        ["year", "player_name", "team", "status", "source"]
    ].head(10)
)
```

```python
sold_players = df_auction[
    (df_auction["status"] == "SOLD") &
    (df_auction["final_price_lakh"].notna())
].copy()

print("Total Sold Players with Price:", len(sold_players))

print("\nAuction Price Statistics (₹ lakh):")

display(
    sold_players["final_price_lakh"].describe()
)

print("\nYear-wise Average Auction Price:")

display(
    sold_players.groupby("year")["final_price_lakh"]
    .agg(["count", "mean", "median", "max"])
    .round(2)
)

print("\nHighest Auction Prices:")

display(
    sold_players[
        ["year", "player_name", "team", "final_price_lakh"]
    ]
    .sort_values("final_price_lakh", ascending=False)
    .head(10)
)
```

```python
import matplotlib.pyplot as plt

yearly_price = (
    sold_players.groupby("year")["final_price_lakh"]
    .mean()
)

plt.figure(figsize=(12, 5))

plt.plot(
    yearly_price.index,
    yearly_price.values,
    marker="o"
)

plt.title("Average IPL Auction Price by Year")
plt.xlabel("Year")
plt.ylabel("Average Auction Price (₹ lakh)")
plt.grid(True)

plt.show()
```

```python
import pandas as pd

df_2024 = pd.read_csv('/content/auction.csv')

print("Dataset Shape:", df_2024.shape)
display(df_2024.head())
```

```python
df_2024_clean = df_2024.rename(columns={
    'PLAYER': 'player_name',
    'TEAM': 'team',
    'PRICE PAID': 'final_price_rupees',
    'TYPE': 'role',
    'NATIONALITY': 'nationality'
})

df_2024_clean['year'] = 2024

df_2024_clean['final_price_lakh'] = (
    pd.to_numeric(df_2024_clean['final_price_rupees'], errors='coerce') / 100000
)

display(df_2024_clean.head())
print("2024 Dataset Shape:", df_2024_clean.shape)
```

```python
import pandas as pd

df_auction = pd.read_parquet('/content/auction_all_years.parquet')

print("Existing Dataset Shape:", df_auction.shape)
print("\nExisting Dataset Columns:")
print(df_auction.columns.tolist())

print("\nExisting Dataset Sample:")
display(df_auction.head())
```

```python
print("Year-wise record count:")
print(df_auction['year'].value_counts().sort_index())

print("\n2024 records in existing dataset:")
display(df_auction[df_auction['year'] == 2024])

print("\n2024 record count:")
print(len(df_auction[df_auction['year'] == 2024]))
```

```python
import pandas as pd

# Copy the existing auction dataset
df_existing = df_auction.copy()

# Prepare 2024 data with matching columns
df_2024_add = df_2024_clean.copy()

# Convert team names to match the existing dataset format
team_mapping = {
    'Chennai Super Kings': 'CSK',
    'Mumbai Indians': 'MI',
    'Kolkata Knight Riders': 'KKR',
    'Royal Challengers Bengaluru': 'RCB',
    'Royal Challengers Bangalore': 'RCB',
    'Rajasthan Royals': 'RR',
    'Delhi Capitals': 'DC',
    'Punjab Kings': 'PBKS',
    'Sunrisers Hyderabad': 'SRH',
    'Gujarat Titans': 'GT',
    'Lucknow Super Giants': 'LSG'
}

df_2024_add['team'] = df_2024_add['team'].replace(team_mapping)

# Add columns required by the existing dataset
df_2024_add['player_id'] = pd.NA
df_2024_add['base_price_lakh'] = pd.NA
df_2024_add['status'] = 'SOLD'
df_2024_add['acquisition_type'] = 'auction'
df_2024_add['source'] = 'manual_2024'

# Keep the same column order
df_2024_add = df_2024_add[df_existing.columns]

# Combine both datasets
df_combined = pd.concat(
    [df_existing, df_2024_add],
    ignore_index=True
)

print("Original dataset:", df_existing.shape)
print("2024 records added:", len(df_2024_add))
print("Combined dataset:", df_combined.shape)

print("\nYear-wise record count:")
print(df_combined['year'].value_counts().sort_index())
```

```python
# Save combined dataset as CSV
df_combined.to_csv(
    '/content/IPL_Auction_2008_2026_Combined.csv',
    index=False
)

print("Combined dataset saved successfully!")

# Verify the saved file
df_check = pd.read_csv('/content/IPL_Auction_2008_2026_Combined.csv')

print("Saved dataset shape:", df_check.shape)
display(df_check.head())
```
